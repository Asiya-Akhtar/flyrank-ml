# **Abstract**

This research investigates whether a **machine-learning model can help identify content matching a defined declining-content condition and prioritize observations for human review**.

The study uses historical records from the **FlyRank ML Internship dataset** and compares several machine-learning approaches with a simple rule-based baseline using the same client-holdout validation split.

The **Random Forest achieved a measured Precision@50 of 0.748 compared with 0.240 for the baseline**, representing approximately **3.08× the baseline performance** on the selected validation split.

The model output is converted into a **ranked content action queue** containing priority scores, reason codes, human-review rules, monitoring considerations, and practical recommendations.

The findings are **directional and decision-support oriented** and should not be interpreted as evidence of causation or as predictions of search-engine behavior.

# **Case Study Framing**

This study addresses a practical **content-prioritization problem**: identifying historical content observations that match a defined declining-content condition and deciding which observations should receive human attention first.

The study uses the **FlyRank ML Internship dataset** as the case-study setting. The goal is **not to predict search-engine behavior or prove why content performance changes**. Instead, the goal is to test whether historical features can help rank observations for review more effectively than a simple baseline.

The resulting workflow is positioned as a **decision-support tool**. It helps a human reviewer decide where to look first, while the final content decision remains with the reviewer.

# **1. Question**

## **Research Question**

**Can a machine-learning model identify content that matches the defined declining-content label and help prioritize which observations should be reviewed first?**

## **Objective**

The objective is to build a **repeatable workflow** that uses historical data to rank observations for human review.

The model is **not intended to predict search-engine behavior, determine the cause of a decline, or automatically change content**.

Instead, it provides a **priority ranking** that helps a reviewer focus attention on higher-priority observations first.

## **Decision Supported**

The main decision supported by this work is:

**Which content observations should a human reviewer investigate first?**

# **3. Methodology**

## **Overall Approach**

The workflow follows these steps:

**1. Prepare the historical data**

**2. Define the declining-content label**

**3. Select eligible features**

**4. Remove potential leakage**

**5. Train several models**

**6. Compare them with a simple baseline**

**7. Evaluate all approaches on the same held-out data**

**8. Select the strongest measured ranking approach**

**9. Convert the output into a ranked action queue**

## **Label Definition**

The target represents whether an observation matches the **defined declining-content condition** in the approved dataset.

This target is treated as a **classification problem**.

Variables used to directly define the target were **not used as model features**.

## **Features**

The model uses historical features that are appropriate for the intended **decision-support task**.

Features that directly reveal the target, contain future-looking information, or could create leakage were excluded.

The intention is that the model should learn from information that could reasonably be available when a reviewer needs to prioritize observations.

## **Baseline**

A simple **rule-based baseline** was used as the reference point.

The baseline provides a straightforward comparison so that the machine-learning models are not judged in isolation.

The key question is whether a model **performs better than the simple baseline under the same evaluation setup**.

## **Models Evaluated**

The workflow compared:

- **Rule-based baseline**
- **Logistic Regression**
- **Decision Tree**
- **Random Forest**

## **Validation Design**

The approaches were evaluated using the **same held-out client-holdout validation split**.

Using the same evaluation data makes the comparison more direct.

A client-level separation helps reduce the risk that observations associated with the same client appear across both training and evaluation sets.

## **Leakage Checks**

Potential leakage was considered during feature selection.

Variables that directly encode the target or use future-looking information were excluded.

The evaluation therefore measures how well the selected features rank observations under the defined validation setup.

The results should be interpreted as **measured model performance**, not as evidence of causation.

# **4. Results**

## **Model vs Baseline**

The approaches were evaluated on the **same client-holdout validation split**.

The primary ranking metric was **Precision@50**.

| **Approach** | **Precision@50** | **ROC-AUC** |
|---|---:|---:|
| **Baseline Rules** | **0.240** | **0.500** |
| **Logistic Regression** | **0.400** | **0.700** |
| **Decision Tree** | **0.540** | **0.742** |
| **Random Forest** | **0.748** | **—** |

## **Key Result**

The **Random Forest achieved the strongest measured Precision@50 at 0.748**.

The baseline achieved **0.240 Precision@50**.

This means the Random Forest's measured Precision@50 was approximately **3.08× the baseline** on the selected validation split.

The Decision Tree achieved **0.540**, while Logistic Regression achieved **0.400**.

Therefore, under this validation setup, the **Random Forest provided the strongest measured ranking performance** among the tested approaches.

## **Interpretation**

The result suggests that the Random Forest was **more effective than the simple baseline at placing observations matching the defined declining-content condition near the top of the review queue**.

However, this is a **measured result on the selected dataset and validation design**.

It does **not** prove:

- **Causation**
- **Search-engine ranking behavior**
- **That the model will perform identically on future data**
- **That changing the recommended content will improve performance**

The result should therefore be described as **directional, measured, and decision-support oriented**.

# **Ranked Output**

The selected **Random Forest model** is used to generate a **ranked priority queue for human review**.

Each observation can be assigned a **priority score** and an associated reason or action recommendation.

The purpose of the queue is to help a reviewer determine **where to investigate first**.

The ranking itself does **not automatically determine what action should be taken**.

A human reviewer should inspect the supporting evidence and content context before making a final decision.

# **5. Limitations and Honest Framing**

This study has several important limitations.

## **Historical Data**

The model learns from **historical observations**. Future conditions may differ from the conditions represented in the training data.

## **No Causal Claim**

The model identifies observations associated with the defined declining-content label.

It does **not** establish why an observation declined or prove that a particular content action would cause future improvement.

## **Feature Limitations**

The model can only use information represented by the available features.

Important factors that are not included in the dataset **cannot be evaluated**.

## **Validation Limitations**

The measured results depend on the selected **validation split, time period, features, and target definition**.

A different dataset or validation design could produce different results.

## **Decision-Support Only**

The system should therefore be treated as a **decision-support workflow**, not an autonomous production system.

A human should **review the evidence before implementing any recommendation**.

# **6. Ranked Recommendations**

The model output is converted into a **ranked action queue**.

The purpose of the queue is to help content reviewers focus their limited time on observations that receive higher priority from the model.

## **Recommended Workflow**

**1. Review the highest-ranked observations first.**

Start with observations receiving the strongest priority scores.

**2. Check the evidence.**

Review the underlying signals and available context before taking action.

**3. Identify the appropriate action.**

Determine whether the observation should be **improved, reviewed, monitored, or handled in another way**.

**4. Apply changes only after human review.**

The model should **not directly publish or modify content**.

**5. Monitor the result.**

After an action is taken, continue observing the relevant signals.

**6. Refresh the workflow when signals become stale.**

If the underlying data changes substantially, the ranking should be reviewed and potentially regenerated.

## **Reason Codes**

Reason codes should make the ranking **understandable to a reviewer**.

Examples include:

- **Declining-content signal**
- **High review priority**
- **Historical performance change**
- **Low relative performance**
- **Needs human investigation**
- **Monitor for further change**

Reason codes provide context for the queue rather than claiming that the model knows the exact cause of a decline.

## **Archetype-to-Action Mapping**

| **Content Situation** | **Possible Action** |
|---|---|
| **Strong performance** | **Protect / Monitor** |
| **Declining performance** | **Review / Improve** |
| **Recovering performance** | **Monitor** |
| **Potentially outdated content** | **Refresh Review** |
| **Weak or unclear evidence** | **Human Investigation** |

These mappings are intended as **decision-support guidance**, not automatic instructions.

## **Decay and Refresh**

Historical signals can become stale.

The action queue should therefore be regenerated when monitoring suggests that the current ranking no longer represents the latest available observations.

## **Human Review**

A reviewer should:

- **Check the evidence behind the ranking**
- **Check the content context**
- **Confirm that the recommended action makes sense**
- **Avoid acting solely because an observation has a high model score**
- **Record important decisions where appropriate**

# **What Should Not Be Automated**

The following decisions should **not be fully automated**:

- **Final content deletion**
- **Major content rewrites**
- **Publishing changes directly to a live website**
- **Decisions based only on a model score**
- **Claims about search-engine causation**
- **Sensitive or ambiguous content decisions**
- **Automatic conclusions about why content declined**

The model should help humans **prioritize investigation**, not replace human judgment.

The system is intended for **practical research and decision support**, not autonomous production control.

# **7. Monitoring / Retrain Triggers**

The recommendations should be monitored because **historical relationships can change over time**.

## **Monitoring Signals**

The workflow should be reviewed when:

- **Model ranking quality decreases**
- **Precision@50 falls materially from the validated result**
- **The distribution of important features changes**
- **The type of content being reviewed changes**
- **The data collection process changes**
- **The definition of the target changes**

## **Refresh Trigger**

A new ranking should be generated when the available historical data has changed enough that the existing queue may no longer represent the current observations.

## **Retraining Trigger**

Retraining should be considered when:

- **New labeled data becomes available**
- **Measured ranking performance declines**
- **Feature distributions shift**
- **The content population changes**
- **The label definition changes**

Retraining should be treated as a **review decision rather than an automatic production process**.
