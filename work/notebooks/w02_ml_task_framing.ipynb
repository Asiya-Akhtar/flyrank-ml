## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is whether a content page **needs to be refreshed**.

Since the dataset does not contain a direct "needs_refresh" label, I would create a proxy based on performance indicators such as declining trend, low CTR, poor average position, or low engagement.

Target values:

- 1 = Needs refresh
- 0 = Does not need refresh

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df["needs_refresh"] = (
    (df["trend_direction"] == "down") &
    (df["avg_position"] > 10)
).astype(int)

df[["content_id", "trend_direction", "avg_position", "needs_refresh"]].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as "refresh every page with low CTR" ignores many other factors that influence content performance.

Machine learning can combine information from multiple features, including search volume, CTR, average position, engagement rate, and trend direction, to identify pages that are most likely to benefit from a refresh. This makes the recommendations more accurate than a single fixed rule.
